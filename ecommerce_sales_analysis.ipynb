## Day 2 – Customer and Order Analysis

```python
unique_orders=sales_data["InvoiceNo"].nunique()
unique_orders
```

```python
unique_customers=sales_data["CustomerID"].nunique()
unique_customers
```

```python
order_revenue=(sales_data.groupby("InvoiceNo")["Revenue"]).sum()
order_revenue
average_order_value=order_revenue.mean()
average_order_value
```

```python
customer_revenue=(sales_data.groupby("CustomerID")["Revenue"].sum().sort_values(ascending=False))
top_customers=customer_revenue.head(10)
top_customers
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    top_customers.index.astype(str),
    top_customers.values
)

plt.title("Top 10 Customers by Revenue")
plt.xlabel("Customer ID")
plt.ylabel("Revenue")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
customer_orders=(sales_data.groupby("CustomerID")["InvoiceNo"].nunique().sort_values(ascending=False))
customer_orders
top_customer_orders=customer_orders.head(10)
top_customer_orders
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    top_customer_orders.index.astype(str),
    top_customer_orders.values
)

plt.title("Top 10 Customers by Number of Orders")
plt.xlabel("Customer ID")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
customer_analysis = pd.DataFrame({
    "Total_Revenue": customer_revenue,
    "Number_of_Orders": customer_orders
})

customer_analysis["Average_Order_Value"] = (
    customer_analysis["Total_Revenue"] /
    customer_analysis["Number_of_Orders"]
)

customer_analysis.head(10)
```

```python
top_aov_customers = (
    customer_analysis
    .sort_values("Average_Order_Value", ascending=False)
    .head(10)
)

print(top_aov_customers)
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    top_aov_customers.index.astype(str),
    top_aov_customers["Average_Order_Value"]
)

plt.title("Top 10 Customers by Average Order Value")
plt.xlabel("Customer ID")
plt.ylabel("Average Order Value")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## Day 3 – Seaborn and Statistical Visualization

```python
import seaborn as sns
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    sales_data["Revenue"],
    bins=50,
    kde=True
)

plt.title("Distribution of Transaction Revenue")
plt.xlabel("Revenue")
plt.ylabel("Number of Transactions")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(x=sales_data["Revenue"])

plt.title("Boxplot of Transaction Revenue")
plt.xlabel("Revenue")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_products.values,
    y=top_products.index
)

plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Product")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

correlation = sales_data[["Quantity", "UnitPrice", "Revenue"]].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=sales_data,
    x="Quantity",
    y="Revenue"
)

plt.title("Quantity vs Revenue")
plt.xlabel("Quantity")
plt.ylabel("Revenue")

plt.tight_layout()
plt.show()
```

## Day 3 – Key Insights

- Seaborn was used to create statistical visualizations from the retail data.
- The transaction revenue distribution is highly right-skewed.
- The boxplot shows the presence of high-value revenue outliers.
- Product-wise revenue was visualized using a Seaborn bar chart.
- Correlation analysis was performed between Quantity, UnitPrice, and Revenue.
- The relationship between Quantity and Revenue was visualized using a Seaborn scatter plot.
- Interactive visualization libraries such as Cufflinks and Plotly were explored, but package compatibility issues prevented their use in this project.
