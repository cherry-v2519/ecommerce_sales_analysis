```python
product_quantity=sales_data.groupby("Description")["Revenue"].sum()
product_quantity
```

```python
product_quantity.sort_values(ascending=False).head(10)
```

```python
top_products = (
    sales_data.groupby("Description")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))
plt.bar(top_products.index, top_products.values)

plt.title("Top 10 Products by Revenue")
plt.xlabel("Product")
plt.ylabel("Revenue")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
sales_data["InvoiceDate"]=pd.to_datetime(sales_data["InvoiceDate"])
```

```python
sales_data.head(20)
```

```python
monthly_revenue = (
    sales_data
    .set_index("InvoiceDate")
    .resample("ME")["Revenue"]
    .sum()
)

plt.figure(figsize=(12, 6))
plt.plot(monthly_revenue.index, monthly_revenue.values)

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
country_revenue=sales_data.groupby("Country")["Revenue"].sum().sort_values(ascending=False).head(10)
country_revenue
```

```python

plt.figure(figsize=(10, 6))

plt.barh(country_revenue.index, country_revenue.values)

plt.title("Top 10 Countries by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.hist(sales_data["Revenue"],bins=50)
plt.title("Distribution of transaction revenue")
plt.xlabel("Revenue")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    sales_data["Quantity"],
    sales_data["Revenue"],
    alpha=0.3
)
plt.title("Quantity vs Revenue")
plt.xlabel("Quantity")
plt.ylabel("Revenue")

plt.tight_layout()
plt.show()
```

```python
unique_orders=sales_data["InvoiceNo"].nunique()
unique_orders
```

```python
unique_customers=sales_data["CustomerID"].nunique()
unique_customers
```

```python
order_revenue=(sales_data.groupby("InvoiceNo")["Revenue"]).sum()
order_revenue
average_order_value=order_revenue.mean()
average_order_value
```

```python
customer_revenue=(sales_data.groupby("CustomerID")["Revenue"].sum().sort_values(ascending=False))
top_customers=customer_revenue.head(10)
top_customers
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    top_customers.index.astype(str),
    top_customers.values
)

plt.title("Top 10 Customers by Revenue")
plt.xlabel("Customer ID")
plt.ylabel("Revenue")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
customer_orders=(sales_data.groupby("CustomerID")["InvoiceNo"].nunique().sort_values(ascending=False))
customer_orders
top_customer_orders=customer_orders.head(10)
top_customer_orders
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    top_customer_orders.index.astype(str),
    top_customer_orders.values
)

plt.title("Top 10 Customers by Number of Orders")
plt.xlabel("Customer ID")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
customer_analysis = pd.DataFrame({
    "Total_Revenue": customer_revenue,
    "Number_of_Orders": customer_orders
})

customer_analysis["Average_Order_Value"] = (
    customer_analysis["Total_Revenue"] /
    customer_analysis["Number_of_Orders"]
)

customer_analysis.head(10)
```

```python
top_aov_customers = (
    customer_analysis
    .sort_values("Average_Order_Value", ascending=False)
    .head(10)
)

print(top_aov_customers)
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    top_aov_customers.index.astype(str),
    top_aov_customers["Average_Order_Value"]
)

plt.title("Top 10 Customers by Average Order Value")
plt.xlabel("Customer ID")
plt.ylabel("Average Order Value")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
